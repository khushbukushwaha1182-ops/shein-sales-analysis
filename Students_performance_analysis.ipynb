```python
#Step2- Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Step3- Read the csv file

df = pd.read_csv('Student_performance_data _.csv')
```

```python
#Step5- View data
#Rows
df.head()
```

```python
#dataset information
df.info()
```

```python
#Column names
df.columns
```

```python
#Step5- Check dataset shape
df.shape
```

```python
#Step6- Basic analysis
#Total students:
len(df)
```

```python
#statistical summary
df.describe()
```

```python
#Step6- Data cleaning
#Check missing values
df.isnull().sum()
```

```python
#Check duplicates
df.duplicated().sum()
```

```python
#Remove duplicates if any
df = df.drop_duplicates()
```

```python
#Step7- KPI analysis
#Total students
total_students = len(df)
print(total_students)
```

```python
#Average GPA
average_gpa = df['GPA'].mean()
print(average_gpa)
```

```python
#Highest GPA
highest_gpa = df['GPA'].max()
print(highest_gpa)
```

```python
#Lowest GPA
lowest_gpa = df['GPA'].min()
print(lowest_gpa)
```

```python
#Average study hours
average_hours = df['StudyTimeWeekly'].mean()
print(average_hours)
```

```python
#Step8- Gender analysis
#Average GPA by gender
average_gpa_by_gender = df.groupby('Gender')['GPA'].mean()
print(average_gpa_by_gender)
```

```python
#Visualization
df.groupby('Gender')['GPA'].mean().plot(kind='bar')
plt.title('Average GPA by Gender')
plt.xlabel('Gender')
plt.ylabel('Average GPA')
plt.show()
```

```python
#Step9= Study time analysis
#Study hours vs GPA
df.groupby('StudyTimeWeekly')['GPA'].mean().head()
```

```python
#Step9= Study time analysis
#Study hours vs GPA
plt.scatter(df['StudyTimeWeekly'], df['GPA'])
plt.title('Study Hours vs GPA')
plt.xlabel('Study Hours')
plt.ylabel('GPA')
plt.show()
```

```python
#Step10- Absence analysis
#Average GPA by analysis
average_gpa_by_gender = df.groupby('Absences')['GPA'].mean()
print(average_gpa_by_gender)
```

```python
plt.scatter(df['Absences'], df['GPA'])
plt.title('Absences vs GPA')
plt.xlabel('Absences')
plt.ylabel('GPA')
plt.show()
```

```python
#Step11- Tutoring analysis
average_gpa_by_gender = df.groupby('Tutoring')['GPA'].mean()
print(average_gpa_by_gender)
```

```python
df.groupby('Tutoring')['GPA'].mean().plot(kind='bar')
plt.title('Tutoring Impact on GPA')
plt.show()
```

```python
#Step12- Grade distribution
df['GradeClass'].value_counts()
```

```python
df['GradeClass'].value_counts().plot(kind='pie', autopct='1%.1f%%')
plt.title('Grade Distribution')
plt.show()
```

```python
#Step13- Correlation Analysis
df.corr(numeric_only=True)
```

```python
#Correlation with GPA
df.corr(numeric_only=True)['GPA'].sort_values(ascending=False)
```
